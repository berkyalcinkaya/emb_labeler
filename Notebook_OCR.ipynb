```python
import re
from rapidocr import RapidOCR
import cv2

_ocr = RapidOCR()

def extract_number(image):
    
    res = _ocr(image)
    if not res or not res.txts:
        return None

    text = " ".join(res.txts)

    m = re.search(r'(\d+(?:\.\d+)?)\s*h', text, re.IGNORECASE)
    if m:
        return float(m.group(1))

    m = re.search(r'\d+(?:\.\d+)?', text)
    return float(m.group()) if m else None
```

```python
timesAll = []

for i in range(1,492):

    image = "F2/F0/D2021.07.05_S00881_I0723_D_WELL3_RUN" + str(i) + ".jpeg" # modify this line - the loops goes through all images
    Img = cv2.imread(image, cv2.IMREAD_GRAYSCALE)
    
    #Imgg = Img[755:799, 755:799] # Use this for 800*800 images - you can automate based on the size, this was just for the test.
    Imgg = Img[455:499, 455:499] # Use this for 500*500 images
    
    aa = extract_number(Imgg)
    timesAll.append(aa)

    print(i, end=" ")
```

```python
import matplotlib.pyplot as plt

plt.plot(timesAll)
```
